```python
# Correlations between different features of dataset
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(),annot=True,cmap='viridis')
```

```python
df['target'].value_counts()
```

```python
# To see visually through a countplot
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('grey')
sns.countplot(x='target',data=df,width=0.5)
plt.xlabel('Target')
plt.ylabel('Count')
plt.title('Countplot of target feature')
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('grey')
sns.countplot(x='sex',hue='target',data=df,width=0.5)
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title("Gender count on the basis of heart disease")
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('grey')
sns.countplot(data=df,x='fbs',hue='target')# fasting biood sugar distribution 
plt.xlabel('Blood Sugar',fontsize=25)
plt.ylabel('Count',fontsize=25)
plt.title('FBS by Targeted People',fontsize=25)
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('grey')
sns.histplot(x='trestbps',data=df,color='black')
plt.show
plt.xlabel('Trestbps',fontsize=25)
plt.ylabel('Count',fontsize=25)
plt.title('Trestbps',fontsize=25)
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('grey')
sns.countplot(x='age',data=df,hue='target')
plt.xlabel('Age',fontsize=25)
plt.ylabel('Count',fontsize=25)
plt.title('Age count by heart disease',fontsize=30)
plt.show
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('black')
sns.countplot(data=df,x='cp',hue='target',palette='CMRmap')
plt.xlabel('Chest Pain',fontsize=25)
plt.ylabel('Count',fontsize=25)
plt.title('Chest Pain by Targeted People',fontsize=25)
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('black')
sns.lineplot(x='cp',data=df,y='age',hue='sex')
#sns.set(rc={'figure.figsize':(15,5)})
plt.xlabel('Chest Pain',fontsize=25)
plt.ylabel('Age',fontsize=25)
plt.title('Chest Pain by Age',fontsize=25)
```

```python
plt.figure(figsize=(12,8))
ax = plt.axes()
ax.set_facecolor('black')
sns.histplot(x='chol',data=df,color='skyblue') # show distribution of serum cholesterol
plt.xlabel('Cholesterol',fontsize=25)
plt.ylabel('Count',fontsize=25)
plt.title('Cholesterol',fontsize=25)
plt.show
```
