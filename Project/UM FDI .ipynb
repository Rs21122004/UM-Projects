```python
# Analysis of investments in sectors from 2016-17
ax = plt.axes()
ax.set_facecolor('cyan')
data['2016-17'].plot(kind='line',figsize=(14,10),title='2016-17',color='black')
plt.gca().spines[['top','right']].set_visible(False)
```

```python
latest_year = data.columns[-1]
investment_latest_year = data[[latest_year,'Sector']].head()
investment_latest_year = investment_latest_year.set_index('Sector')
investment_latest_year.plot(kind='bar',figsize=(12,8),color='skyblue')
plt.title('Investment distribution across sectors for the latest years')
plt.xlabel('Sector')
plt.ylabel('Investment')
plt.xticks(rotation=45)
plt.legend(latest_year)
plt.show()
```

```python
corr_matrix = data.iloc[:,1:].corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix,annot=True,cmap='viridis',fmt='.2f')
plt.title('Correlation matrix of investment across years')
plt.show()
```
